 ## INDUSTRY: HealthyLife Pharmacy.
#### A drug inventory management system  


## TASK 1

```python
# Creating the Variables 
drug_name = "Paracetamol"
price = 850
quantity_in_stock = 250
quantity_sold_today = 65
minimum_stock = 100
is_expired = False
```

```python
print (drug_name)
```

## TASK 2

```python
# Calculating Remaining Stock
remaining_stock = quantity_in_stock - quantity_sold_today
print("Remaining Stock:", remaining_stock)
```

## TASK 3

```python
# Calculating Sales Revenue 
sales_revenue = price * quantity_sold_today
print("Sales Revenue: ₦", sales_revenue, )
```

## TASK 4

```python
# Updating the stock quantity using the subtraction assignment operator
quantity_in_stock -= quantity_sold_today
print("Updated Stock:", quantity_in_stock)
```

## TASK 5

```python
# Evaluates to True if stock is below minimum, False otherwise
needs_restocking = quantity_in_stock < minimum_stock
print("Needs Restocking:", needs_restocking)
```

## TASK 6

```python
# Safe to sell only when it has not expired
is_safe_to_sell = not is_expired 
print("Safe to Sell:", is_safe_to_sell)
```

## TASK 7

```python
# Must be above minimum stock AND safe to sell
final_approval = (quantity_in_stock > minimum_stock) and is_safe_to_sell
print("Final Approval:", final_approval)
```

## TASK 8

```python
# Qualifies if stock > 200 OR revenue > 50,000
eligible_for_discount = (quantity_in_stock > 200) or (sales_revenue > 50000)
print("Eligible for Discount:", eligible_for_discount)
```

## TASK 9

```python
# DRUG INVENTORY SUMMARY
print("========== DRUG INVENTORY REPORT ==========")
print("Drug Name:", drug_name)
print("Price per Unit:", "₦",price)
print("Current Stock:", quantity_in_stock, "units" )
print("Units Sold Today:", quantity_sold_today)
print("Today's Revenue:", "₦", sales_revenue) 
print("Needs Restocking:", needs_restocking)
print("Expired:", is_expired)
print("Safe to Sell:", is_safe_to_sell)
print("Eligible for Discount:", eligible_for_discount)
print("===========================================")
```

### Inventory Management Comparison

```python
# 1. Create variables for Amoxicillin
amox_name = "Amoxicillin"
amox_price = 1500
amox_stock = 90
amox_sold = 35
amox_min_stock = 80
amox_expired = True

# 2. Perform the calculations
amox_revenue = amox_price * amox_sold
amox_stock -= amox_sold # Update inventory
amox_needs_restock = amox_stock < amox_min_stock
amox_safe_to_sell = not amox_expired

# 3. Output Amoxicillin Summary
print("---", amox_name, "Summary---")
print("Current Stock:", amox_stock, "units")
print("Today's Revenue:", "₦", amox_revenue)
print("Needs Restocking:",amox_needs_restock)
print("Safe to Sell:", amox_safe_to_sell, "\n" )

# 4. Compare both drugs (Answering via Boolean logic since 'if' statements are banned)
print("--- Drug Comparison ---")

# Which generated more revenue?
paracetamol_more_revenue = sales_revenue > amox_revenue
print("Paracetamol generated more revenue than Amoxicillin:", paracetamol_more_revenue)

# Which should be restocked?
print("Paracetamol needs restocking:", needs_restocking)
print("Amoxicillin needs restocking:", amox_needs_restock)

# Which should not be sold? (Checking if NOT safe to sell)
print("Paracetamol should NOT be sold:", not is_safe_to_sell)
print("Amoxicillin should NOT be sold:", not amox_safe_to_sell)

# Which is safer for customers? (Checking which one is safe to sell)
paracetamol_is_safer = is_safe_to_sell and (not amox_safe_to_sell)
print("Paracetamol is the safer drug for customers right now:", paracetamol_is_safer)
```
